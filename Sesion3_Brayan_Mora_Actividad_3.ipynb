# **Problema: ¿te alcanza para el café?**

```python
# Definimos las variables

presupuesto = 300000
precio_cafe = 7200
cafes_semana = 10
```

```python
# Hacemos los cálculos

# Gasto al mes

gasto_mes = (precio_cafe * cafes_semana)*4

# Cuanto sobra

sobra = presupuesto - gasto_mes

# ¿Me alcanza?

if sobra >= 0:
  alcanza = True
else:
  alcanza = False
```

```python
# Imprimimos resultados

print(gasto_mes, sobra, alcanza)
```
