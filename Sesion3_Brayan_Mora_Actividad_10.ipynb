# **La caja de la tienda**

```python
# Se crea la lista de precios
precios = [10000, 5000, 3500, 17000]
```

```python
# 1. Suma el total con un for
Total=0

for n in precios:
  Total = Total + n

print(f"El total es: {Total}")
```

```python
# 2. Si el total pasa de 500,
#    aplica 10 % de descuento

if Total>500:
  Descuento=Total*0.1
  Total_con_descuento=Total-Descuento

print(f"El total con descuento es: {Total_con_descuento}")
```

```python
# 3. Imprime el ticket con f-strings:
# Total: ___
# Descuento: ___
# A pagar: ___

print(f"Ticket de compra: \nTotal: {Total} \nDescuento: {Descuento} \nA pagar: {Total_con_descuento}")
```
