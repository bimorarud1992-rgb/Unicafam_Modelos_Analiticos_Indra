**03 · TALLER DE PYTHON · BUCLE WHILE**

# **Tu meta de ahorro**

```python
# Definimos variables
ahorro = 1000  # ← lo que ya tengo
meta = 100000
por_mes = 10000
meses = 0
```

```python
# Mientras no llegues a la meta:
# suma lo del mes y cuenta un mes más

while ahorro<meta:
  ahorro=ahorro+por_mes
  meses+=1
  print(f"El ahorro del mes {meses} es de {ahorro}")
```

```python
# ¿Cuántos meses tardaste?

print(f"Tardaste {meses} meses en completar tu ahorro")
```
