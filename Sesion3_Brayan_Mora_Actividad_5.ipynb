# **Tu lista del súper**

```python
# Definimos la lista
lista = ["Galletas", "Panela", "Arroz", "Sal"]
```

```python
# 1. Imprimimos el primero y el último

print(f"Primer producto: {lista[0]}")
print(f"Último producto: {lista[3]}")
```

```python
# 2. ¿Cuántos productos son?

print(f"La cantidad de productos es: {len(lista)}")
```

```python
# 3. Agrega uno más y vuelve a contar

lista.append("Azúcar")
print(f"La cantidad de productos es: {len(lista)}")
```
