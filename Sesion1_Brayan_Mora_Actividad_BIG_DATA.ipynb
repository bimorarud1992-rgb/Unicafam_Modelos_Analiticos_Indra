# **Ejercicio 3. Primer análisis de ventas**

```python

import pandas as pd
import numpy as np
```

```python

# Creamos números aleatorios
np.random.seed(42)
n = 200
```

```python

#Generamos la tabla de datos
ventas = pd.DataFrame({
      "fecha": pd.date_range("2026-01-01", periods=n, freq="D"),
      "ciudad": np.random.choice(["Bogotá", "Medellín", "Cali", "Barranquilla"], n),
      "canal": np.random.choice(["Tienda", "Web", "App"], n),
      "unidades": np.random.randint(1, 20, n),
      "precio": np.random.choice([25000, 48000, 99000], n),
  })
```

```python
#Agregamos la columna "Total" a la tabla que creamos anteriormente
ventas["total"] = ventas["unidades"] * ventas["precio"]
```

```python
#Imprimimos la tabla (resultados)
print(ventas.shape)
print(ventas.head(3))
```

```python
#Tabla dinámica
resumen = (ventas.groupby("canal")["total"]
                 .agg(["count", "sum", "mean"])
                 .sort_values("sum", ascending=False))
```

```python
#Resumen de los datos
resumen["sum"] = (resumen["sum"] / 1e6).round(1)
resumen["mean"] = resumen["mean"].round(0)
resumen.columns = ["ventas", "total_millones", "ticket_prom"]
```

```python
#Total vendido por canal
print(resumen)
```

```python
#Tabla dinámica
resumen2 = (ventas.groupby("ciudad")["total"]
                 .agg(["count", "sum", "mean"])
                 .sort_values("sum", ascending=False))
```

```python
#Resumen de los datos
resumen2["sum"] = (resumen2["sum"] / 1e6).round(1)
resumen2["mean"] = resumen2["mean"].round(0)
resumen2.columns = ["ventas", "total_millones", "ticket_prom"]
```

```python
#Ticket promedio por ciudad
print(resumen2)
```

```python
#Tabla dinámica
resumen3 = (ventas.groupby(ventas["fecha"].dt.day_name())["total"]
                 .agg(["count", "sum", "mean"])
                 .sort_values("sum", ascending=False))
```

```python
#Resumen de los datos
resumen3["sum"] = (resumen3["sum"] / 1e6).round(1)
resumen3["mean"] = resumen3["mean"].round(0)
resumen3.columns = ["ventas", "total_millones", "ticket_prom"]
```

```python
#Ventas por día
print(resumen3)
```
