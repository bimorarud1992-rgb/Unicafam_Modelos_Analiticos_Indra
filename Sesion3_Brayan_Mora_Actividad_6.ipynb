# **¿Te toca envío gratis?**

```python
# Definimos variables

total = 1241
es_miembro = True
compras = 4
```

```python
# 1. Envío: más de 1000 → gratis;
#    miembro → gratis; si no → $99

if total>1000:
  print("Por la cantidad comprada, tu envío es gratis")
elif es_miembro:
  print("Por ser miembro, tu envío es gratis")
else:
  print("Tu envío cuesta: $99")
```

```python
# 2. Nivel: 10 o más → Oro;
#    de 5 a 9 → Plata; menos → Bronce

if compras>=10:
  print("Eres nivel Oro")
elif compras>=5:
  print("Eres nivel Plata")
else:
  print("Eres nivel Bronce")
```
