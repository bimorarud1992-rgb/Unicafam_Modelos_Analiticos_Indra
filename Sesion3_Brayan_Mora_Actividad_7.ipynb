**03 · TALLER DE PYTHON · BUCLE FOR**

# **Revisa tu ticket de compra**

```python
# Definimos los precios de compra
precios = [200, 550, 100, 1200]
```

```python
# 1. Imprime cada precio con un for

for n in precios:
  print(n)
```

```python
# 2. Suma el total
Total=0

for n in precios:
  Total = Total + n

print(f"El total es: {Total}")
```

```python
# 3. ¿Cuántos cuestan más de 100?

Caros = 0

for n in precios:
  if n>100:
    Caros+=1

print(f"Tienes {Caros} productos caros")
```
